# Get the most information as possible without looking at the data in the csv

```python
# Get the most information as possible without looking at the data in the csv
cumulative_print = ""

def separate_prints(cumulative_print):
    return cumulative_print + "--------------------------------\n"

# Columns names
cumulative_print = separate_prints(cumulative_print)
cumulative_print += str(diabetes_df.columns.tolist())

# Data types
cumulative_print = separate_prints(cumulative_print)
cumulative_print += str(diabetes_df.dtypes)

# Describe the data
cumulative_print = separate_prints(cumulative_print)
cumulative_print += str(diabetes_df.describe())

# Check for missing values
cumulative_print = separate_prints(cumulative_print)
cumulative_print += str(diabetes_df.isnull().sum())

# Check for duplicates
cumulative_print = separate_prints(cumulative_print)
cumulative_print += str(diabetes_df.duplicated().sum())
```
